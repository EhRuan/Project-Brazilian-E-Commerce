# Project-Brazilian-E-Commerce

O projeto tem como base uma análise de dados utilizando o conjunto de dados do Brazilian E-Commerce, <a href="https://www.kaggle.com/datasets/olistbr/brazilian-ecommerce">link de acesso</a>, onde ocorrerá uma limpeza dos dados com correções, criação de dashboards para uma visualização fácil do que foi feito e tratamento futuros

# Importação das bibliotecas e carregamento dos arquivos csv

```python
# Importação das bibliotecas

import pandas as pd
import numpy as np
```

```python
# Carregamento dos arquivos .csv para as variáveis definidas para análises
 
df_customers = pd.read_csv("Dataset/archive/olist_customers_dataset.csv")
df_geolocation = pd.read_csv("Dataset/archive/olist_geolocation_dataset.csv")
df_items = pd.read_csv("Dataset/archive/olist_order_items_dataset.csv")
df_payments = pd.read_csv("Dataset/archive/olist_order_payments_dataset.csv")
df_reviews = pd.read_csv("Dataset/archive/olist_order_reviews_dataset.csv")
df_orders = pd.read_csv("Dataset/archive/olist_orders_dataset.csv")
df_products = pd.read_csv("Dataset/archive/olist_products_dataset.csv")
df_sellers = pd.read_csv("Dataset/archive/olist_sellers_dataset.csv")
df_category = pd.read_csv("Dataset/archive/product_category_name_translation.csv")

# Um dicionário que engloba todas as tabelas
tabelas = {
       "customers": df_customers, "geolocation": df_geolocation,
       "items": df_items, "payments": df_payments, "reviews": df_reviews,
       "orders": df_orders, "products": df_products,
       "sellers": df_sellers, "category": df_category,
}
```

# Visualizações dentro das tabelas

```python
# Visualiza as primeiras 5 linhas das tabelas

for nome, df in tabelas.items():
    print(f"--- Tabela {nome.capitalize()} ---")
    print(df.head())
    print('=-='*25)
```

```python
# Verifica as informações de cada base de dados como os tipos e valores nulos dentro de cada tabela

for nome, df in tabelas.items():
    print(f"--- Tabela {nome.capitalize()} ---")
    df.info()
    print()
```

# Pré-processamento de dados

## Conversão para data

```python
# Converte as colunas de data em formato de texto para o formato datetime

colunas_data = {
    "orders": ["order_purchase_timestamp", "order_approved_at",
               "order_delivered_carrier_date", "order_delivered_customer_date",
               "order_estimated_delivery_date"],
    "items": ["shipping_limit_date"],
    "reviews": ["review_creation_date", "review_answer_timestamp"],
}

for nome, colunas in colunas_data.items():
    for col in colunas:
        tabelas[nome][col] = pd.to_datetime(tabelas[nome][col])
```

```python
# Confere se as colunas de data foram convertidas para datetime

for nome, colunas in colunas_data.items():
    print(f"--- {nome} ---")
    print(tabelas[nome][colunas].dtypes)
```

## Nulos e Duplicatas em todas as tabelas

```python
# Exibe a quantidade de nulos e de linhas duplicadas de cada tabela

# Percorre o dicionário
for nome, df in tabelas.items():
    print(f"--- {nome} ---")
    print("Nulos por coluna:")
    print(df.isnull().sum()[df.isnull().sum() > 0])
    print("Linhas totalmente duplicadas:", df.duplicated().sum())
    print()
```

## Nulos em orders por status

```python
# Cruzando valores nulos de orders com order_status

colunas_nulas = ["order_approved_at", "order_delivered_carrier_date",
                 "order_delivered_customer_date"]

for col in colunas_nulas:
    print(f"--- Nulos em {col} por status ---")
    print(df_orders[df_orders[col].isnull()]["order_status"].value_counts())
    print()
```

## Duplicatas pela chave

```python
# Verifica duplicatas pela chave de cada tabela (o esperado é 0, exceto onde indicado)

print("orders (order_id):", df_orders["order_id"].duplicated().sum())
print("customers (customer_id):", df_customers["customer_id"].duplicated().sum())
print("products (product_id):", df_products["product_id"].duplicated().sum())
print("sellers (seller_id):", df_sellers["seller_id"].duplicated().sum())
print("items (order_id + order_item_id):",
      df_items.duplicated(subset=["order_id", "order_item_id"]).sum())
print("payments (order_id + payment_sequential):",
      df_payments.duplicated(subset=["order_id", "payment_sequential"]).sum())
print("reviews (review_id):", df_reviews["review_id"].duplicated().sum())
```

## Pedidos entregues com inconsistência

```python
# Confere os pedidos 'delivered' com datas nulas (inconsistências de registro)

delivered = df_orders[df_orders["order_status"] == "delivered"]
print("Sem data de aprovação:", delivered["order_approved_at"].isnull().sum())
print("Sem data de envio:", delivered["order_delivered_carrier_date"].isnull().sum())
print("Sem data de entrega:", delivered["order_delivered_customer_date"].isnull().sum())
```

```python
# Separa os pedidos entregues com data de entrega válida, para análises de prazo

df_entregues = df_orders[(df_orders["order_status"] == "delivered") &
                         (df_orders["order_delivered_customer_date"].notnull())].copy()
print(df_entregues.shape)
```

## Tratamento de geolocation

```python
# Agrupa geolocation por prefixo de CEP, usando a média das coordenadas

df_geolocation = (
    df_geolocation
    .groupby("geolocation_zip_code_prefix")
    .agg(geolocation_lat=("geolocation_lat", "mean"),
         geolocation_lng=("geolocation_lng", "mean"),
         geolocation_city=("geolocation_city", "first"),
         geolocation_state=("geolocation_state", "first"))
    .reset_index()
)

# Atualiza o dicionário, pois df_geolocation foi reatribuída
tabelas["geolocation"] = df_geolocation
```

## Tratamento de products

```python
# Corrige o erro de digitação ("lenght") e preenche categorias nulas

df_products = df_products.rename(columns={
    "product_name_lenght": "product_name_length",
    "product_description_lenght": "product_description_length",
})
df_products["product_category_name"] = df_products["product_category_name"].fillna("desconhecida")

# Atualiza o dicionário, pois df_products foi reatribuída
tabelas["products"] = df_products
```

## Conferência final

```python
# Confere o tamanho e a quantidade de colunas por tipo após o tratamento

for nome, df in tabelas.items():
    print(f"--- {nome} ---")
    print(f"{df.shape[0]} linhas, {df.shape[1]} colunas")
    print(df.dtypes.value_counts())
    print()
```
